***Add a new note***: Append text to notes.txt.

***View all notes***: Read and display all notes from notes.txt.

```python

def view_notes():
    try:
      with open("note.txt", "r") as file:
        notes = file.readlines()
        if notes:
           print("\n-----Your Notes-----")
           for i, note in enumerate(notes):
              print(f"{i + 1}. {note.strip()}")
           print("-----------------------")
        else:
             print("No notes found.")
    except FileNotFoundError:
         print("No note file found, Start by adding a note!")
```

```python
print("Welcome to your simple note taker!")

while True:
  print("Choose an option: ")
  print("\n 1. Add a note")
  print("\n 2. View all notes")
  print("\n 3. Exit")

  choice = input("Enter your choice (1/2/3): ")

  if choice == '1':
    note_content = input("Enter your note: ")
    add_note(note_content)
  elif choice == '2':
    view_notes()
  elif choice == '3':
    print("Exiting Note Taker. GoodBye! ")
    break  # to terminate the while loop immediately
  else:
    print("Invalid choice. Please try again.")
```
