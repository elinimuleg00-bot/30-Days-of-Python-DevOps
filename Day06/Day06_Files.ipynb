```python
# Using with statement (recommended)
# This comment indicates that the code uses a 'with' statement, which is a best practice for handling files in Python.
with open("diary.txt", "w") as file:
    # This line opens a file named 'diary.txt'.
    # The 'w' mode means it's opened for writing. If the file doesn't exist, it will be created.
    # If it does exist, its contents will be truncated (emptied).
    # The 'as file' part assigns the opened file object to a variable named 'file'.
    # The 'with' statement ensures the file is automatically closed, even if errors occur.
    file.write("2023-07-20: Learned file handling!\n")
    # This line writes the string "2023-07-20: Learned file handling!" followed by a newline character (\n) into the 'diary.txt' file.
    file.write("2023-07-21: Built a file organizer\n")
    # This line writes another string, "2023-07-21: Built a file organizer", also followed by a newline, to the same 'diary.txt' file, appending it after the previous line.
```

```python
#Reading file
with open("diary.txt", "r") as file:
  # The 'r' mode opens an existing file so you can read or extract its data without changing
    content = file.read()
    print(content)
```

```python
# Reading lines
with open("diary.txt", "r") as file:
    lines = file.readlines() # gives you the actual content of each line in a list
    print("Journal entries:", len(lines)) # tells you how many lines there are in total.
```

```python
# Writing to a file
file = open("example.txt", "w")
#The open() function returns a file object, which is then assigned to the variable file.
file.write("First line\nSecond line\n")
file.close()
# Failing to close a file can lead to unexpected behavior or loss of data.Doing it saves any changes
```

```python
# Reading from a file
file = open("example.txt", "r")
content = file.read()
print("File content:")
print(content)
file.close()
```

```python
import csv
# built-in csv module provides functionality to work with CSV(Comma Separated Values) files
# makes it easier to read from and write to them without manually handling commas and quotes.

# Writing CSV
with open("data.csv", "w", newline='') as file:
  # newline='' : ensures that there are no extra blank rows in the CSV output.
    writer = csv.writer(file)
  # It takes the file object (the one we just opened) as an argument, so the writer knows where to write the CSV data.
    writer.writerow(["Name", "Age", "City"])
  # writerow() is a method of the writer object that takes a list of strings (or other data types that can be converted to strings)
  # and writes them as a single row in the CSV file, separated by commas.
  # the list beacomes the column headers (first row) to the file
    writer.writerow(["Alice", 28, "New York"])
  # the line wries the second row of data
    writer.writerow(["Bob", 32, "London"])
```

```python
# Reading CSV
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader: # loop through each row (list of strings)
        print(f"{row[0]} | {row[1]} | {row[2]}")
        # in an embeded expressin of f-string get the first.second and third item sequencely
```
