**Regex**(**Regular** **Expression**)

Normally, when you press Ctrl+F and search for a word(e.g: apple), it only finds the exact word. However Regex allow you to search patterns instead of specific words.

```python
import re # importing python's built-in Regular Expression Module

# Basic Patterns
text = "The quick brown fox"  # defines a string variable to search through
match = re.search(r"brown", text) # scans text from left to right for literal substring 'brown' using r - raw string. Returns Match object if found, or none if not found.
print("Found brown:" , bool(match))  # prints a message, converts the returned Match object into boolean(True - if found, False - if None)


# Character set: used to parse command flags, file permissions, or custom log severity level
vowels = re.findall(r"[aeiou]", text, re.I)
      # Square bracket [] define a Character Set, matches any single character that is either a,e,i,o,or u
      # re.I - case-insesitive flag without modifying the string (matches both lowercase and uppercase)
      # re.findall - scans the entire string and returns all the matching charcters
print("Vowels:", vowels)


# Quantifiers: used to extract port numbers(0-1023,...), HTTP status codes, and timestamps out of server logs
numbers = "123 4567 89012" # a string containing whitespace-separated number of varying lengths string
print("3-4 digits:", re.findall(r"\d{3,4}", numbers))
 #\d -> any single numerical digit(0-9), {3,4} -> range quantifier matches length between 3 and 4 digits long  since we r using \d it helps us how long the character is, but \d+ won't help it will just take the whole character
 # during scanning everything will pass except the 5 digits, so python takes the max allowed chuck of digits only

```

 **When** **to** **use** **which**:

 - *re*.*search* () - when u only care if a pattern exists once, or u just want to extract the first occurnence.
 - *re*.*findall* () - when a line or file might have a multiple target items and u want to extract all of them.


```python
import re

# Word boundaries
text = "cat category catapult wildcat" # words that starts with cat and ends
print("Whole words: " , re.findall(r"\bcat\b", text)) #output: cat in cat ONLY. and ignores others
print("Whole words: " , re.findall(r"cat\b", text)) #output: cat in wildcat, and cat itself only. Ignores others bcuz it was followed by a word
# \b (word boundary) -> matches the boundary between a word character and a non-word character(spaces,punctuation, or start/end of line)
# the whole r"" tells regex to match cat only when it stands alone as a complete word.


# Digit matching
log = "Error 404: Not found"
error_code = re.search(r"\d+", log) # \d+ match one or more consecutive digits
print("Error code:", error_code.group()) #.group - reaches inside the Match object and pulls the string out


# Negative Lookahead(with a d\t style of writing)
print("Cat not followed by 'alog' : ",
      re.findall(r"cat(?!alog)", "catalog cat"))
#(?!alog) - the negative lookahead that only match 'cat' if it is not followed by 'alog'
```

```python
import re

# Capturing Groups
date = "2023-12-25"
match = re.match(r"(\d{4})-(\d{2})-(\d{2})", date)
# re.match - checks if the pattern matches starting from the beginning of the string
# (\d{4}) -> Group 1: Captures 4 digits ,   - -> matches the literal hyphen
print(f"Year: {match.group(1)}, Month: {match.group(2)}") # .group(0) would return the entire


# Named Groups: does the same exact thing link capturing groups but instead of referencing by index numbers(0,1,2), u give them a descriptive labels
text = "James: 42"
match = re.search(r"(?P<name>\w+): (?P<age>\d+)", text)
# ?P<name> -> labels name to this group,  \w+ -> matches one or more word character, capturing James
print(f"{match.group('name')} is {match.group('age')}")


# Non-Capturing Groups
print("Plurals:", re.findall(r"\b\w+(?:s|es)\b", "boxes cats dog"))
#\b -> ensures the pattern starts and ends at a word boundary,
#(s|es) -> matches the suffix s and es and ONLY returs as ['es', 's'] , but the (?:...) forces it to treturn the whole words
```

```python
import re

#Simple Substitution(re.sub)
text = "Order: 12345, Status: pending"
redacted = re.sub(r"\d{5}", "XXXXX", text) # replacing any 5 digits into XXXXX
print("Redacted:", redacted)


# Replacement function
def stars(match): # a custom function that recieves a regex Match object
  return '*' * len(match.group()) # generates '*' of the same exact length of that matched string

sensetive = "Card: 4111-1111-1111-1111"
secured = re.sub(r"\d{4}(-\d{4}){3}", stars, sensetive)  #\d{4} -> matches 4 starting digits
# (-\d{4}){3} -> matches a hyphen followed by 4 digits, repeated exactly 3 times
# after it finds the digits, it passes that match into stars(), and then stars counts 19 characters and generates * by swapping the digits
print("Secured: ", secured)
```
