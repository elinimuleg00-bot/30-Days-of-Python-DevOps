Shopping List Manager (CLI Utility)
-----------------------------------
A simple interactive script demonstrating lists, functions,
while-loops, and input sanitization in Python.

```python
shopping_list = []

def add_item(item):
  shopping_list.append(item.lower())
  print(f"{item} added to your list." )

def remove_item(item):
  if item.lower() in shopping_list:
     shopping_list.remove(item.lower())
     print(f"{item} removed from your list." )
  else:
    print(f"{item} not found in your list.")

def display_list():
  if not shopping_list:
    print("Your list is empty.")
  else:
    print("\n --- Your Shopping List ---")
 # enumerate() loops you through a list while taking care of the index automatically
 # i will take the index while item will take the actual item from shopping_list
 # the second argument 1 tells it to start counting from 1 and not 0 (default)
    for i, item in enumerate(shopping_list, 1):
      print(f"{i}. {item.capitalize()}")
      print("-----------------------------")
```

```python
print("Welcome to your Shopping List Manager!")
print("Type 'add' to add an item, 'remove' to remove an item, 'display' to view your list, 'quit' to exit")

while True:
  action = input("What would you like to do? ").lower()

  if action == 'add':
    item_to_add = input("Enter the item to add: ")
    add_item(item_to_add)
  elif action == 'remove':
    item_to_remove = input("Enter the item you want to remove: ")
    remove_item(item_to_remove)
  elif action == 'display':
    display_list()
  elif action == 'quit':
    print("Goodbye!")
    break
  else:
    print("Invalid command. Please try again. ")


```
