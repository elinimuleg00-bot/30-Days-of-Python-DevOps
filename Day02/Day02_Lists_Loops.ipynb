```python
# Implicit conversion
print(3 + 2.5)  # int + float = float

# Explicit conversion
num_str = "123"
num_int = int(num_str)
print(num_int * 2)
```

```python
# List creation and basic methods - mutable
fruits = ["apple", "banana"]
print(f"Original list: {fruits}")

# Adding elements
fruits.append("orange")
fruits.insert(1, "mango")
print(f"After additions: {fruits}")

# Removing elements
fruits.remove("banana")
popped = fruits.pop()
#The popped variable is 'orange' because the pop() method, when called without an argument, removes and returns the last item from the list
print(f"After removals: {fruits}, Popped: {popped}")

# Changing elements
fruits[0] = "Passion Fruit"
print(f"After changes: {fruits}")

# Sorting
numbers = [3, 1, 4, 1, 5]
numbers.sort(reverse=True)
print(f"Sorted numbers: {numbers}")
```

```python
# Slicing
numbers = [0, 1, 2, 3, 4, 5]
# extracts the elements from index 2 up to (but not including) index 4
print("Middle elements:", numbers[2:4])
print("Last two:", numbers[-2:])

# List comprehension - the sequence will be 1, 2, 3, 4, 5 whereas each of these numbers x is then squared (x**2) to form the squares list.
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

# Combining lists
list1 = [1, 2, 3]
list2 = [4, 5]
combined = list1 + list2
print(f"Combined: {combined}")
```

```python
# Basic tuple - immutable
dimensions = (1920, 1080)
width, height = dimensions    # tuple unpacking allowing to assign the individual elements of tuple in a single line
print(f"Resolution: {width}x{height}")

# Multiple return values
def get_user_info():  # a function without any arguments
    return "John", "Doe", 30

first, last, age = get_user_info()
print(f"{first} {last}, {age} years old")

# Extended unpacking
values = (1, 2, 3, 4, 5)
a, b, *rest = values
print(f"First: {a}, Second: {b}, Rest: {rest}")
```

```python
# Basic for loop
for i in range(5):
    print(i)

# List iteration
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I love {fruit}s")
```

```python
# Basic while loop
count = 0
while count < 5:
    print(count)
    count += 1

# Break statement
while True:
    user_input = input("Enter 'quit' to exit: ")
    if user_input.lower() == 'quit':
        break
```

## Mini Project: Shopping List

This simple program will allow you to manage a shopping list using concepts we've explored: lists, functions, loops, and user input.

```python
shopping_list = []

def add_item(item):
    shopping_list.append(item.lower())
    print(f"'{item}' added to the list.")

def remove_item(item):
    if item.lower() in shopping_list:
        shopping_list.remove(item.lower())
        print(f"'{item}' removed from the list.")
    else:
        print(f"'{item}' not found in the list.")

def display_list():
    if not shopping_list:
        print("Your shopping list is empty.")
    else:
        print("\n--- Your Shopping List ---")
        for i, item in enumerate(shopping_list, 1):
            print(f"{i}. {item.capitalize()}")
        print("--------------------------")
```

```python
print("Welcome to your Shopping List Manager!")
print("Type 'add' to add an item, 'remove' to remove an item, 'display' to view your list, or 'quit' to exit.")

while True:
    action = input("\nWhat would you like to do? ").lower()

    if action == 'add':
        item_to_add = input("Enter the item to add: ")
        add_item(item_to_add)
    elif action == 'remove':
        item_to_remove = input("Enter the item to remove: ")
        remove_item(item_to_remove)
    elif action == 'display':
        display_list()
    elif action == 'quit':
        print("Goodbye!")
        break
    else:
        print("Invalid command. Please try again.")
```
