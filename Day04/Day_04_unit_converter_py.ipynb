# Unit Converter

A lightweight interactive conversion script demonstrating functions with multi-parameter handling, input normalization, conditional routing, and fallback error handling in Python.

```python
def unit_converter(value, from_unit, to_unit):

  """
  Converts a value from one unit to another.

  Args:
    value(float): The value to convert.
    from_unit(str): The unit to convert from.
    to_unit(str): The unit to convert to.

  Returns:
    float: The converted value.
  """

  from_unit = from_unit.lower() # 'Kg' will be treated the same as 'kg'
  to_unit = to_unit.lower()

  if from_unit =='kg' and to_unit =='lbs':
    return value * 2.20462
  elif from_unit == 'lbs' and to_unit == 'kg':
    return value / 2.20462
  elif from_unit == 'cm' and to_unit == 'in':
    return value * 0.393701
  elif from_unit == 'in' and to_unit == 'cm':
    return value  / 0.393701
  elif from_unit == to_unit:
    return value # no conversion is needed
  else:
    print(f"Unsupported conversion: {from_unit} to {to_unit}")
    return None
```

```python
# Convert 10 kg to lbs
kg_value = 10
lbs_result = unit_converter(kg_value, 'kg', 'lbs')
if lbs_result is not None:  # acts as a safeguard returns none for unsupported conversions
  print(f"{kg_value} kg is equal to {lbs_result:.2f} lbs")

# Convert 50 cm to inches
cm_value = 50
inches_result = unit_converter(cm_value, 'cm', 'in')
if inches_result is not None:
  print(f"{cm_value} cm is equal to {inches_result:.2f} inches")

# Convert 20 lbs to kg
lbs_value = 20
kg_result = unit_converter(lbs_value, 'lbs', 'kg')
if kg_result is not None:
  print(f"{lbs_value} lbs is equal to {kg_result:.2f} kg")

# Convert 15 inches to cm
in_value = 15
cm_result = unit_converter(in_value,'in', 'cm')
if cm_result is not None:
  print(f"{in_value} inches is equal to {cm_result:.2f} cm")

# Example of unsupported conversion
unit_converter(10, 'kg', 'miles')
```
