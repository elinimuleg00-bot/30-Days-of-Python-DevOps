# Simple Calculator
A lightweight interactive script demonstrating arithmetic operations, user input handling, numerical type casting, and functional evaluation in Python.




```python
def add(x,y):
  return x + y

def subtract(x,y):
  return x - y

def multiply(x,y):
  return x * y

def division(x,y):
  if y == 0:
    return "Error! no division by 0"
  else :
    return x / y

def power(x,y):
  return x**y


print("Select operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Division")
print("5. Power")


while True:
  choice = input ("Enter Choice (1/2/3/4/5)")
  if choice in ('1','2','3','4','5'):
    num1 = float(input("Enter first number:"))
    num2 = float(input("Enter second number:"))

    if choice == '1':
      print(num1 , "+", num2 , "=", add(num1,num2))
    elif choice == '2':
      print(num1 , "-", num2, "=", subtract(num1,num2))
    elif choice == '3':
      print(num1, "x", num2 , "=", multiply(num1, num2))
    elif choice == '4':
      print(num1, "/" , num2 , "=", division(num1,num2))
    elif choice == '5':
      print(num1 , "***", num2, "=", power(num1,num2))

      next_calcluation = input("Let's do next caclulation? (yes/no) :")
      if next_calcluation.lower() == "no" :
        break

      else:
        print("Invalid Input !")

```
