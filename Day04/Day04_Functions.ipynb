```python
def format_name(first, last):
    """Formats a full name from first and last name"""
    return f"{first.title()} {last.title()}"

names_to_format = [
    ("alice", "smith"),
    ("charlie", "brown"),
    ("diana", "prince")
]

formatted_names = []  # an empty list to store the names after they got formatted
for first, last in names_to_format: # in each iteration, it unpacks the tuple in names_to_format:
                                    # the first element of the tuple assigned to first
                                    # and the second element to last
    formatted_name = format_name(first, last) # for each first and last name pair it calls the function
    formatted_names.append(formatted_name) # adds the formatted name to the empty list so that we prevent reassigning

print("Formatted Names:")

for name in formatted_names:
   print(name)
```

```python
# Default parameters
def calculate_area(length, width=1): # default parameter is the width, whie calling
                                     # if the width is not assigned it automatically
                                     # use a default value of 1
    """Calculates area (defaults to square if width omitted)"""
    return length * width
calculate_area(10)

# you can also call like this
# print(calculate_area(width=4, length=5))  Order doesn't matter
```

```python
# Variable-length arguments (*args)
def average(*numbers):  # the functipn can accept any number of positional arguments
    """Calculates average of any number of values"""

    return sum(numbers) / len(numbers) if numbers else 0
# sum(numbers) - if numbers is not empty calculates the sum of all elements in the numbers tuple
# len(numbers) - the number if elements in the tuple numbers
# sum(numbers) / len(numbers) - performs average
# else 0 - if the numbers is empty, it returns 0 avoiding a ZeroDivisionError

print(f"Average: {average(4, 7, 9, 2):.2f}")
# :. 2f farmat s it to 2 dp
```

```python
# Simple return
def square(number):
    return number ** 2

result = square(4)
print(f"Square of 4: {result}")

# Returning multiple values, including a middle name
def split_name(full_name):
    names = full_name.split()

    if not names: # Handles empty string
        return None, None, None
    elif len(names) == 1: # Handles single name (e.g., "Cher")
        return names[0], None, None
    elif len(names) == 2: # Handles first and last name (e.g., "Marie Curie")
        return names[0], None, names[1]
    else: # Handles first, middle(s), and last name (e.g., "Marie Sklodowska Curie")
        return names[0], ' '.join(names[1:-1]), names[-1]

        # names[1:-1] - creates a new list containing elements from the names list,
   # starting from index 1 (the second element) up to, but not including, the last element (-1)
        # ' '.join(...) - This is a string method that takes an iterable (like a list of
  # strings) as its argument and concatenates all the elements into a single string

# Example 1: Two names
first, middle, last = split_name("Marie Curie")
print(f"Two names: First: {first}, Middle: {middle}, Last: {last}")

# Example 2: Three or more names (with a middle name)
first, middle, last = split_name("Marie Sklodowska Curie")
print(f"Three names: First: {first}, Middle: {middle}, Last: {last}")

# Example 3: Single name
first, middle, last = split_name("Cher")
print(f"Single name: First: {first}, Middle: {middle}, Last: {last}")

# Example 4: Empty string
empty_first, empty_middle, empty_last = split_name("")
print(f"Empty string: First: {empty_first}, Middle: {empty_middle}, Last: {empty_last}")
```

```python
# Returning complex data
def analyze_number(n):
    is_even = n % 2 == 0 # checks if n is an even, and assign it to is_even variable thats a boolean value comparing results ==0
    absolute = abs(n) # calculates the absolute value of the input using the built-in abs()
    return {  # returning a dictionary(collection of key-value pairs)
        "original": n,
        "even": is_even,
        "absolute": absolute
    }

print(analyze_number(-5))

# accessing the specific value in dictionary
result_dict = analyze_number(10)
print(f"Original number: {result_dict['original']}")
print(f"Is even: {result_dict['even']}")
print(f"Absolute value: {result_dict['absolute']}")
```
