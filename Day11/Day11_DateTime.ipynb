```python
import pandas as pd

data = {
    "timestamp_str" : ["2026-09-01 08:15:00", "2026-09-02 14:30:00", "2026-09-03 22:45:00"],
    "server_load_pct": [42.1, 88.5, 12.3]
}
df = pd.DataFrame(data)

#converting string column to Pandas datetime(A structured object holding calendar components {Year-Month-Day Hour:Minute:Second}.)
df["timestamp"] = pd.to_datetime(df["timestamp_str"]) # the variable assigned creates a new column named timestamp in my df

print("Data Type:\n", df.dtypes)


#extracting time components for ML
df["year"] = df["timestamp"].dt.year # extracts year using .dt.year from our column
df["month"] = df["timestamp"].dt.month
df["day"] = df["timestamp"].dt.day
df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek  # Monday = 0, Sunday = 6
df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)
# .isin([5, 6]): Checks if the number in day_of_week is equal to 5 (Saturday) or 6 (Sunday). This returns a boolean column of True or False.
# .astype(int): Converts True into 1 and False into 0 because machine learning models require numerical inputs instead of boolean words.

print("\n--- ML-READY DATETIME FEATURES ---")
print(df[["timestamp", "hour", "day_of_week", "is_weekend"]])


#filtering & sorting by date range
cutoff_date = "2026-09-02 00:00:00" # sets a boundary
recent_logs = df.loc[df["timestamp"] >= cutoff_date]
# df.loc[...]: Keeps only the rows where the check evaluated to True

print(f"\n--- LOGS ON OR AFTER {cutoff_date} ---")
print(recent_logs)
```
