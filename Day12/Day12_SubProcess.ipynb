The **subprocess** module acts as a bridge between Python and your system's underlying OS shell.

```python
# Basic System Command
import subprocess

result = subprocess.run(["ls","-l"], capture_output=True, text=True)
# ["ls", "-l"] : command passed as a list of strings (where "ls" is the command to list files, and "-l" is the flag for the long detailed format showing file sizes, dates, and permissions). Passing arguments as a list prevents command-injection security vulnerabilities.
# capture_output=True: Captures both standard output (stdout) and error messages (stderr) from the terminal so you can save them into a Python variable instead of letting them dump directly into the console.
# text=True: Automatically decodes the system's raw binary output (bytes) into a readable Python text string (str).
# result: The variable storing a CompletedProcess object returned by subprocess.run(){executes a terminal command}.

print("Command Output:\n", result.stdout)
```

```python
import urllib.request # for sending web requests across the internet without depending on external system binaries like ping

try:
    # Test network connectivity by opening a lightweight HTTP connection. Sets a maximum wait limit of 3 seconds. If Google doesn't respond within 3 seconds (e.g., if internet access is disabled),
    # Python stops waiting and raises a timeout error
    response = urllib.request.urlopen("https://google.com", timeout=3)
    print(f"Network Check: Success! HTTP Status Code: {response.getcode()}")
except Exception as e:
    print(f"Network Check: Failed! Error: {e}")
```

```python
# MLOps Use Case - Memory Check

import subprocess

result = subprocess.run(["free", "-h"], capture_output = True, text = True)
# "free": A Linux terminal command that reports total, used, and available system memory (RAM and Swap space).
# "-h": Stands for "Human-readable" format. Instead of printing memory in raw bytes (like 16384920), it formats values cleanly in megabytes or gigabytes (like 15Gi).


print("--- System Memory Report ---")
print(result.stdout)
```
