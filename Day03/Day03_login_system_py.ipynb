# Simple Login System



A simple interactive authentication script demonstrating conditional logic, string comparison operators, and user input validation in Python.

```python
expected_username = "admin"
expected_password = "password123"

user_input_username = input("Please enter your username: ")
user_input_password = input("Please enter your password: ")

if user_input_username == expected_username and user_input_password == expected_password:
  print("Login Successfully !")
else:
  print("Login Declined !")
```
