# Folder Scanner
A Python automation script demonstrating OS directory traversal using os.listdir(), path validation with os.path.exists(), and file vs. directory classification using os.path.join().

```python
import os

# the parameter represents the folder path I want to search
def scan_folder(directory_path):
  print(f"--- Scanning Dircetory: {directory_path} ---")

# asks the operating system: "Does this folder pathway actually exist on the hard drive?" If it does not exist, this statement evaluates to True.
  if not os.path.exists(directory_path):
    print("Error: That directory does not exist!")
    return # stops the function so Python doesn't try to scan a non-existent folder and crash

# goes to the target folder and gathers a list of everything inside it(filnames and nested folder names) and save it to string 'items'
  items = os.listdir(directory_path)

  if not items: # if the list is empty
    print("The folder is empty.")
    return

  for item in items: # a loop pulling out every single file and folder name from items list and assign it each into item

  # glues the folder path and the specific filename together
     full_path = os.path.join(directory_path,item)

     if os.path.isfile(full_path):  # checks if its a file
        print(f"File: {item}")
     elif os.path.isdir(full_path): # checks if its a directory(nested folder)
        print(f"Folder: {item}")

scan_folder(".")  # executes the function
                    # "." tells to look at the current working directory

# prints .config (where Colab stores setup settings) and sample_data (a folder Google provides by default with CSV files).
```

# File Organizer
An automated directory management utility demonstrating file extension parsing with os.path.splitext(), dynamic directory creation using os.makedirs(), and physical file relocation via shutil.move().

```python
import os # to read filenames and make new folders
import shutil # physically cut and paste the files into their new homes


# Create temporary messy files to organize
dummy_files = ["test1.txt", "test2.txt", "photo.jpg", "vacation.jpg", "script.py"]

for filename in dummy_files:
    with open(filename, "w") as f:
        f.write("This is a test file.")

print("Created 5 test files successfully!")


def organize_files(target_directory):
  print(f"--- Organizing Files in: {target_directory} ---")

  # opens the target folder,looks at everything inside, and starts looping through every single item
  for item in os.listdir(target_directory):
    full_path = os.path.join(target_directory, item)
    if os.path.isdir(full_path):
      continue  # stop running the current loop cycle and skip straight to the next item. We do this because we only want to sort loose files—we don't want to try moving folders inside other folders

  # takes a filename string and cuts it into a tuple of two halves right at the final dot "photo.jpg --> photo, .jpg" and then tuple unpacking them
    file_name, file_extension = os.path.splitext(item)

 # takes the file extension replacing that period(.) into ntg (e.g: .jpg -> jpg) and then into uppercase (JPG)
    ext_folder_name = file_extension.replace(".","").upper()

    if not ext_folder_name:
      continue # This if statement catches files that has no extension at all so it doesn't break the script

# target_directory contains "." while ext_folder contaings "JPG" so it joins them using backslash(/) such as -  "./JPG"
    new_folder_path = os.path.join(target_directory, ext_folder_name)

    if not os.path.exists(new_folder_path): # checks if the specific extension folder exists
      os.makedirs(new_folder_path) # if not exist yet, build a new folder directory right then and there
      print(f"Created Folder: {ext_folder_name}")

#For example, moving "photo.jpg" into the "JPG" folder means the final address is "./JPG/photo.jpg".
    destination_path = os.path.join(new_folder_path, item)

    shutil.move(full_path, destination_path)
    print(f"Moved: {item} -> {ext_folder_name}/")

organize_files(".")
```
