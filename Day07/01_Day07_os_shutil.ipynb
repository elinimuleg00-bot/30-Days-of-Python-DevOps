```python
import os # tells Python to load the built-in Operating System module. It allows your script to talk directly to whatever computer or server is running it (like Google's servers when you are in Colab).
import shutil # loads the Shell Utilities module. It gives Python extra powers to handle high-level file operations like copying, moving, or checking hardware storage.

print("---SYSTEM CHECKS ---")
#Path Handling
#stands for "Get Current Working Directory." It asks the system: "What folder is this script running inside right now?" * It saves that folder path into a variable called current_path. (In Google Colab, this will usually print /content).
current_path = os.getcwd()
print(f"1. Working Directory: {current_path}") # prints out the folder path

# holds the exact name of the file we want to look for(built on Day 6)
file_to_check = "notes.txt"

# os.path.exists - function that physically goes out to the hard drive to look for the file name and if its gotten, evaluates 'True'
if os.path.exists(file_to_check):
  print(f"2. Status: Found your 'file_to_check' file!")
else:
  print(f"2. Status: '{file_to_check}' does not exist here yet.")

# Hardware Check(Checking Disk Space)
# shutil.disk_usage("/") : function looks at your hard drive storage. The "/" means the root directory (the entire drive).
# Also uses tuple unpacking, returns 3 separate numbers at the exact same time: the total capacity, the amount used, and the amount free.
total, used, free = shutil.disk_usage("/")

#The computer counts storage in bytes, which is massive, unreadable numbers
# so I am converting it to GB
free_gb = free / (1024**3)
print(f"3. Storage: You have {free_gb:.2f} GB of free space left in Colab. ")

```
