```python
# Dataframe Creation
# Imports the Pandas( used for data manipulation, analysis, and cleaning)library into Python. The as pd part creates a standard industry alias/shortcut so you can type pd instead of pandas
import pandas as pd

#From Dictionary
data = {
    "Name": ["Alice", "Bob","Charlie"],
    "Age": [25,30,35],
    "City": ["Paris", "London","Berlin"]
}
df = pd.DataFrame(data) # Converts the dictionary into a DataFrame(Panda's 2D table object like a spreadsheet)
print("From Dictionary:\n", df)

# From lists of lists
data = [
    [100, "Widget A", 19.99],
    [101, "Widget B", 29.99],
    [102, "Widget C", 39.99]
]
# since raw list doesn't have keys to use as column names, i pass the columns parameter to define it
df = pd.DataFrame(data,columns=["ID", "Product", "Price"])
print("\nFrom Lists:\n", df)
```

```python
#csv opertaions
import pandas as pd
data = {
    "Temperature": [22.1,23.5,19.8],
    "Humidity": [45,62,38]
}
# Write to csv
df = pd.DataFrame(data)
df.to_csv("weather.csv",index=False) #saves df to a new file named weather.csv on ur disk, index = false is crucial preventing pandas from writing the automatic row numbers(0,1,2) into csv (but shows index in output bcuz df is called)

#read from csv
df = pd.read_csv("weather.csv") # reads it, which brings that data back into python's memory as df, during writing the data was out to ur hardrive
print("CSV Data:\n", df)

# append to csv
new_data = pd.DataFrame({"Temperature":[21.3],"Humidity":[55]})
new_data.to_csv("weather.csv", mode = "a", header = False, index = False) #mode = a -> opens the existing file in Append mode(instead of the default 'w', which wipes and overwrite), header = false ->the CSV file already has column headers at the top so pandas skips writing a header
updated_df = pd.read_csv("weather.csv")
print("Updated CSV Data(4 rows):\n", updated_df)
```

```python
# Data Selection
import pandas as pd
df = pd.read_csv("weather.csv")

# Column Selection
print("Humidity Column:\n", df["Humidity"]) # extracts a single column by its header, returns a single vertical column with row index

# Row Filtering
print("\nHigh Humidity:\n", df[df["Humidity"] > 50]) #Wraps the boolean mask inside df[]. Pandas keeps only the rows where the condition evaluated to True

# Integer-Based Indexing
print("\nFirst Row:\n", df.iloc[0])
#df.iloc(Integer Location) selects rows and columns purely by their index, so [0] pulls out the first row of data

#Label & Conditional Location
print("\nTemp > 20:\n", df.loc[df["Temperature"] > 20 ])
# df.loc(Label Location)- selects data based on explicit row/column labels or conditional boolean checks
```

**When** **to** **Use:**
- .*iloc* : When u know the numerical position(e.g: give me the first 5 rows will look like -> df.iloc[0:5])
- .*loc* : When u filter by conditional rules or specific column names

```python
# Data Cleaning
import pandas as pd
import numpy as np # provides np.nan (Not a Number), the standard IEEE floating-point representation for missing or null data values.

# Create sample data with missing values
data = {
    "A": [1, 2, np.nan, 4],
    "B": ["X", np.nan, "Z", "W"],
    "C": [0.1, 0.2, 0.3, np.nan]
}
df = pd.DataFrame(data)

# Handle missing data
print("Original: \n", df)
print("\nDrop NA: \n", df.dropna()) # Scans the DataFrame and removes any row that contains at least one NaN value in any column.
print("\nFill NA: \n", df.ffill())
# df.fillna(...): Replaces missing values with real data instead of dropping rows.
# method="ffill": Stands for "Forward Fill". It takes the value from the previous valid row and copies it down into the missing slot.

# Type conversion
df["A"] = df["A"].astype("Int64") # Standard int64 fails when NaN is present. so we r allowing it to exist alongside NaN without converting to float
print("\nFixed Types:\n", df.dtypes) # prints data type for each column
```
