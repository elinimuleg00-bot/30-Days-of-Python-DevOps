# ML Dataset Cleaner & Feature Scaler
A foundational machine learning data-preprocessing script demonstrating tabular data loading with Pandas, median imputation for missing value handling (fillna), feature standardization using scikit-learn's StandardScaler, and automated CSV export.

```python
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler # ML library in python, that rescales big and small numbers into a balanced range

# setting up a file & fake data wiz missing values
CSV_FILE_PATH = "raw_ml_dataset.csv"  # The uppercase letters signal to developers: 'This is a fixed setting (constant)'
raw_data = """age,income,experience_years,target
25,50000,2,0
30,,5,1
,85000,10,1
45,120000,,0
50,110000,20,1
"""

with open(CSV_FILE_PATH, "w")as file:
  file.write(raw_data)

def clean_and_scale_dataset(file_path):
  try:
    df = pd.read_csv(file_path)
    print("--- 1. RAW DATASET (WITH MISSING VALUES) ---")
    print(df)

    #fixing missing values(Imputation)
    numeric_cols = ["age", "income", "experience_years"] # list containing only columns we want to fix and scale leaving target

    # Calculate medians so you can see what is being inserted
    medians = df[numeric_cols].median()
    print("\n--- CALCULATED COLUMN MEDIANS ---")
    print(medians)
     # df[numeric_cols].median(): Calculates the exact middle value (median) for each numeric column.

    df[numeric_cols] = df[numeric_cols].fillna(medians)
    print("\n--- 2. DATASET AFTER MEDIAN IMPUTATION ---")
    print(df)
    # fillna(...): finds every missing cell and fills it with that column's median number
    # df[numeric_cols] = ... : Overwrites the original columns with the newly repaired data.

    # future scaling (Standardization)
    scaler = StandardScaler() # df[numeric_cols] = ...: Overwrites the original columns with the newly repaired data.
    df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
    print("\n--- 3. FINAL CLEANED & SCALED ML DATASET ---")
    print(df)
    # scaler.fit_transform(...): Does two things in one step:fit: Calculates the average and spread (standard deviation) of the numbers in each column.transform: Re-calculates every single number using the formula Z = (x - mean)/ the spread

    # Exporting Clean Data & Error Handling
    output_path = "cleaned_ml_dataset.csv"
    df.to_csv(output_path, index=False)
    print(f"\n✅ Success! Saved to '{output_path}'")

  except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
  except Exception as e:
    print(f"An unexpected error occured: {e}")

if __name__ == "__main__":
  clean_and_scale_dataset(CSV_FILE_PATH)
```
