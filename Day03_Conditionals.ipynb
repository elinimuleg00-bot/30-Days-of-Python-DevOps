```python
score = 85

# elif
if score >= 90:
    grade = "A"
elif score >= 80 & score <= 90:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Your grade: {grade}")
```

```python
age = 25
has_membership = True

#nested if-else
if age >= 18:
    print("Adult verified")
    if has_membership:
        print("Welcome premium member!")
    else:
        print("Regular adult access")
else:
    print("Minor access only")
```
