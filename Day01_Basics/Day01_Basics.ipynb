### Using Variables with Multi-line Strings
You can embed variables directly into multi-line strings using f-strings. This allows for dynamic content within your multi-line text blocks.

```python
name = "Alice"
age = 30

message = f"""Hello, {name}!
You are {age} years old.
This is a multi-line message with variables."""

print(message)
```

```python
# Using single quotes with explicit newlines
single_quote_multiline = 'This is a line.\nThis is another line.'
print(single_quote_multiline)

print("\n---\n") # Separator for clarity

# Using double quotes with explicit newlines
double_quote_multiline = "This is one line.\nAnd this is the next line."
print(double_quote_multiline)
```

```python
# Basic input
name = input("Enter your name: ")
print(f"Hello, {name}!")

# Numeric input
age = int(input("Enter your age: "))
print(f"You'll be {age + 5} in 5 years")
```
