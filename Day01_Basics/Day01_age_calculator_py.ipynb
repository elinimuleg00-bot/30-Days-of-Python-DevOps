```python
name = input("Enter your name: ")

#note we use int() because input() always gives us text(string)
birth_year = int(input("Enter your birth year: "))

current_year = 2026
age = current_year - birth_year

print(f"Hello {name}! Based on the year {current_year}, you are {age} years old.")
```
