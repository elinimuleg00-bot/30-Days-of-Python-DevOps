The **finally** **block** acts as the ultimate cleanup crew. No matter what happens inside the try or except blocks—whether the code succeeds, fails, or encounters an unexpected error—the code inside finally will **ALWAYS** **run**.

```python
try:
  file = open("data.txt", "r")
  content = file.read()
  print(content) # attempts to open the file and print the text, if not found it triggers an error
except FileNotFoundError:
  print("File doesn't exist")
finally:  # executes this block at the very end no matter what
  file.close() if 'file' in locals() else None  #  checks if the variabble file was actually created, if the file opened successfully, it calls file.close() to free up space
                                                #  if opening file failed, avoids .close() to prevent a new crash thats why "if 'file' in locals()" is there
  print("cleanup complete")

  # when opening a file its better to close it when u r done, unless that file can stay locked or corrupted in system memory
```

**Raising** **Exception** : to trigger your own errors on purpose by using the keyworrd 'raise'. this helps us catch the error quick before it reaches our main code

```python
def validate_age(age):
  if age < 0:
    raise ValueError("Age cannot be negative") # stop execution and throw an error, triggering a standard ValueError with a custom message

  if age > 120:
    raise ValueError("Unrealistic age")

try:
    validate_age(-5)
except ValueError as e: # catches an error and saves its message into e
    print(f"Invalid age: {e}")
```

Custom Exception

```python
class InvalidEmailError(Exception):  # this class inherits all powers from the standard Exception parent class, the try except block is programmed to only catch errors that belong to the Exception family. and the exception has the ability to hold a message
  pass # tells python that this class doesn't need extra code just having the name is enough

def validate_email(email):
  if "@" not in email:
    raise InvalidEmailError("Missing @ symbol")

try:
  validate_email("usergmail.com")
except InvalidEmailError as e:
  print(f"Email Error: {e}")
```
