# Safe File Reader & Error Handler
A robust file-handling script demonstrating defensive I/O operations, explicit try-except-finally blocks for specific exception management, resource cleanup, and UTF-8 encoding verification in Python.

```python
import os

def safe_read_file(file_path):
  file_handle = None  # gives finally block a reference, without an error

  try:
    print(f"---Attempting to open: {file_path} ---")

# opens a file in a read mode and an encoding(tells a computer how to convert binary numbers into human-readable characters)
# if the file is missing, an error occurs and this block handles it by stopping the execution and jumping into the except block
    file_handle = open(file_path, "r", encoding = "utf-8")

# reads the file content and prints it out
    content = file_handle.read()
    print("\n File Content: \n" + content)

#handles diffrent error types with custom message
  except FileNotFoundError:
    print(f" Error: The file '{file_path}' was not found. Please check the path.")

  except IsADirectoryError:
    print(f"Error: '{file_path}' is a directory folder, not a file.")

  except PermissionError:
    print(f"Error: Access denied. You don't have permission to read '{file_path}'.")

  except UnicodeDecodeError:
    print(f"Error: Could not read '{file_path}'. It may be a binary file (like an image or zip).")

  except Exception as e:
    print(f"Unexpected Error: An unkown error occured: {str(e)}")

# ensures system resources are cleaned up while preventing crashes if opening the file failed in the first place.
# by checking if the file was successfully opened and is still open before calling .close()
  finally:
    if file_handle is not None and not file_handle.closed:
      file_handle.close()
      print("System Cleanup: File resource successfully closed.")
    else:
      print("System Cleanup: No open file resource to close.")



      #==============================================
        # Testing the script under diffrent conditions
      #==============================================

# Test 1: File doesn't exist
 # calling my function, passing a filename that doesn't exist on the hard drive
safe_read_file("non_existent_file.txt")

print("\n" + "="*40 + "\n")  # visually divides Test 1 from Test 2

# Test 2: File exists and reads successfully
with open("sample.txt", "w") as f: # the with ststement closes automatically as soon as the block finishes
  f.write("Hello! This is a successful file read test.")

safe_read_file("sample.txt")
```
