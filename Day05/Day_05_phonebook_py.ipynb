# Phonebook Manager
A simple interactive script demonstrating dictionary data structures (key-value mapping), modular functions, and basic CRUD (Create, Read, Delete) operations in Python.




```python
phonebook = {}

def add_contact(phonebook_dict, name, number):
  #temporary, local name for the dictionary within the function's scope
  phonebook_dict[name]=number
  print(f"Contact '{name} with number '{number}' added.")

#Example
add_contact(phonebook,"Alice", "101-938-837-8289")
add_contact(phonebook, "Bob", "938-383-558-8382")
print("Current PhoneBook: " , phonebook)
```

```python
def find_contact(phonebook_dict, name):
  if name in phonebook_dict:
    print(f"Contact '{name}' found with number '{phonebook_dict[name]}'.")
  else:
    print(f"Contact '{name}' not found.")

#Example
find_contact(phonebook, "Alice")
find_contact(phonebook, "Charlie")
```

```python
def delete_contact(phonebook_dict, name):
  if name in phonebook_dict:
    del phonebook_dict[name]
    print(f"Contact '{name} deleted.'")
  else:
    print(f"Contact '{name}' not found in phonebook.")

#Example
delete_contact(phonebook, "Alice")
delete_contact(phonebook, "Charlie")
print("Current PhoneBook after deletion: ", phonebook)
```
